```python
# renombrar columnas
df_IV = df_IV.rename(columns={
    "Peso\n(g) \n": "Peso_g",
    "Longitud (cm)": "Longitud_cm",
    "p H": "pH",
    "Oxígeno disuelto (OD)": "OD",
    "Sólidos totales ppt": "TDS",
    "Salinity ‰": "Salinidad",
    "T° C Aire": "Temp_Aire",
    "T° H2O": "Temp_Agua",
    "Lux": "Lux"
})
df_MAR = df_MAR.rename(columns={
    "# Especie": "Especie",
    "Longitud cm": "Longitud_cm",
    "Peso (g)": "Peso_g",
    "OD (ppm)": "OD",
    "Conductividad (ms/cm)": "Conductividad",
    "TDS (ppt)": "TDS",
    "Salinity  ‰": "Salinidad",
    "T°C Aire": "Temp_Aire",
    "T°C H2O": "Temp_Agua",
    "Lux": "Lux"
})
df_T = df_T.rename(columns={
    "Peso (g)": "Peso_g",
    "longitud (cm)": "Longitud_cm",
    "Oxígeno disuelto": "OD",
    "Sólidos totales (ppt)": "TDS",
    "% salinidad": "Salinidad",
    "T°C Aire": "Temp_Aire",
    "T° H2O": "Temp_Agua"
})
```

```python
#  Seleccionar solo las columnas comparables entre los tres sistemas
cols_comunes = [
    "Especie", "Peso_g", "Longitud_cm", "VIVA=1 MUERTA=0",
    "Temp_Aire", "Temp_Agua", "pH", "OD", "Conductividad",
    "TDS", "Salinidad", "Lux", "Sistema"
]
```

```python
# Añadir columna “Sistema”
df_IV["Sistema"] = "INVITRO"
df_T["Sistema"] = "TANQUE"
df_MAR["Sistema"] = "MAR"
```

```python
# Crear los dataframes filtrados
df_IV2 = df_IV[cols_comunes]
df_T2 = df_T[cols_comunes]
df_MAR2 = df_MAR[cols_comunes]
```

```python
#Unir los tres sistemas
df_total = pd.concat([df_IV2, df_T2, df_MAR2], ignore_index=True)
```

```python
df_total
```

# 1. ANOVA 3 sistemas (Peso y Longitud)

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Asegurar que Especie y Sistema sean categóricas
df_total["Especie"] = df_total["Especie"].astype("category")
df_total["Sistema"] = df_total["Sistema"].astype("category")

# ANOVA Peso ~ Especie * Sistema
modelo_peso = smf.ols("Peso_g ~ Especie * Sistema", data=df_total).fit()
anova_peso = sm.stats.anova_lm(modelo_peso, typ=2)
print("ANOVA Peso_g")
print(anova_peso)

# ANOVA Longitud ~ Especie * Sistema
modelo_long = smf.ols("Longitud_cm ~ Especie * Sistema", data=df_total).fit()
anova_long = sm.stats.anova_lm(modelo_long, typ=2)
print("\nANOVA Longitud_cm")
print(anova_long)
```

```python
# EMMeans Especie × Sistema
import itertools


# Cuadrícula Especie × Sistema
niveles_especie = df_total["Especie"].cat.categories
niveles_sistema = df_total["Sistema"].cat.categories

grid = pd.DataFrame(
    list(itertools.product(niveles_especie, niveles_sistema)),
    columns=["Especie", "Sistema"]
)

# EMMeans Peso
pred_peso = modelo_peso.get_prediction(grid).summary_frame()
emmeans_peso = grid.copy()
emmeans_peso["EMM_Peso"] = pred_peso["mean"]
emmeans_peso["EMM_Peso_IC95_inf"] = pred_peso["mean_ci_lower"]
emmeans_peso["EMM_Peso_IC95_sup"] = pred_peso["mean_ci_upper"]

# EMMeans Longitud
pred_long = modelo_long.get_prediction(grid).summary_frame()
emmeans_long = grid.copy()
emmeans_long["EMM_Long"] = pred_long["mean"]
emmeans_long["EMM_Long_IC95_inf"] = pred_long["mean_ci_lower"]
emmeans_long["EMM_Long_IC95_sup"] = pred_long["mean_ci_upper"]

# Unir en una sola tabla
emmeans_total = (
    emmeans_peso
    .merge(emmeans_long, on=["Especie","Sistema"])
)

print(emmeans_total)
```

## Gráficas comparativas 3 sistemas (EMMeans + IC95%)

```python
#  EMMeans de Peso por Especie × Sistema

# Convertir a categoría (soluciona el error)
emmeans_total["Especie"] = emmeans_total["Especie"].astype("category")
emmeans_total["Sistema"] = emmeans_total["Sistema"].astype("category")

plt.figure(figsize=(10,6))

sns.pointplot(
    data=emmeans_total,
    x="Especie",
    y="EMM_Peso",
    hue="Sistema",
    dodge=0.3,
    errorbar=None
)

# Obtener categorías una sola vez
cats = list(emmeans_total["Especie"].cat.categories)

# Añadir IC95% manualmente
for i, fila in emmeans_total.iterrows():
    x_pos = cats.index(fila["Especie"])
    offset = {"INVITRO": -0.2, "TANQUE": 0, "MAR": 0.2}[fila["Sistema"]]
    plt.plot(
        [x_pos + offset, x_pos + offset],
        [fila["EMM_Peso_IC95_inf"], fila["EMM_Peso_IC95_sup"]],
        color="gray",
        linewidth=1.5
    )

plt.title("EMMeans de Peso por Especie y Sistema (IC95%)")
plt.ylabel("Peso ajustado (EMM)")
plt.xlabel("Especie")
plt.legend(title="Sistema")
plt.tight_layout()
plt.show()


```

```python
# EMMeans de Longitud por Especie × Sistema
plt.figure(figsize=(10,6))

sns.pointplot(
    data=emmeans_total,
    x="Especie",
    y="EMM_Long",
    hue="Sistema",
    dodge=0.3,
    errorbar=None
)

for i, fila in emmeans_total.iterrows():
    x_pos = list(emmeans_total["Especie"].cat.categories).index(fila["Especie"])
    offset = {"INVITRO": -0.2, "TANQUE": 0, "MAR": 0.2}[fila["Sistema"]]
    plt.plot(
        [x_pos + offset, x_pos + offset],
        [fila["EMM_Long_IC95_inf"], fila["EMM_Long_IC95_sup"]],
        color="gray",
        linewidth=1.5
    )

plt.title("EMMeans de Longitud por Especie y Sistema (IC95%)")
plt.ylabel("Longitud ajustada (EMM)")
plt.xlabel("Especie")
plt.legend(title="Sistema")
plt.tight_layout()
plt.show()
```

```python
#ANOVA 3 sistemas (Peso y Longitud)
import statsmodels.api as sm
import statsmodels.formula.api as smf

df_total["Sistema"] = df_total["Sistema"].astype("category")
df_total["Especie"] = df_total["Especie"].astype("category")

modelo_peso = smf.ols("Peso_g ~ Sistema * Especie", data=df_total).fit()
anova_peso = sm.stats.anova_lm(modelo_peso, typ=2)
print(anova_peso)

modelo_long = smf.ols("Longitud_cm ~ Sistema * Especie", data=df_total).fit()
anova_long = sm.stats.anova_lm(modelo_long, typ=2)
print(anova_long)
```

Peso correlaciona fuerte con salinidad, OD y temperatura.

Longitud correlaciona moderadamente con salinidad y temperatura.

pH y Lux tienen efectos menores.

Las gráficas ambientales muestran que el crecimiento en peso y longitud está fuertemente asociado a la salinidad, la temperatura del agua y el oxígeno disuelto. El sistema marino presenta valores más altos de estas variables, lo que se traduce en un crecimiento significativamente mayor, especialmente en la especie A.s. El sistema tanque muestra condiciones intermedias y un crecimiento moderado, mientras que el sistema in vitro, con salinidad baja, temperatura controlada y oxígeno disuelto reducido, produce el menor crecimiento. Las correlaciones y regresiones confirman que la salinidad es el predictor más fuerte del crecimiento, seguido por la temperatura del agua y el oxígeno disuelto.
