```python
df_IV['Especie'] = df_IV['Especie'].astype(str)
df_IV['Tratamiento'] = df_IV['Tratamiento'].astype(str)
```

```python
df_IV['Grupo'] = df_IV[['Especie','Tratamiento']].agg('_'.join, axis=1)
```

```python
# Post-hoc Tukey HSD (Peso y Longitud)
# Peso
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_peso = pairwise_tukeyhsd(
    endog=df_IV['Peso_g'],
    groups=df_IV['Grupo'],
    alpha=0.05
)

print(tukey_peso)

```

```python
# # Post-hoc Tukey HSD ( Longitud)
tukey_long = pairwise_tukeyhsd(
    endog=df_IV['Longitud_cm'],
    groups=df_IV['Grupo'],
    alpha=0.05
)

print(tukey_long)
```

```python
# letras de significatividad
    

def tukey_to_letters(tukey_result):
    # Extraer comparaciones
    groups = list(tukey_result.groupsunique)
    letters = {g: "" for g in groups}

    # Crear matriz de significancia
    sig_matrix = pd.DataFrame(
        np.ones((len(groups), len(groups))), 
        index=groups, 
        columns=groups
    )

    # Rellenar con 0 cuando hay diferencia significativa
    for i in range(len(tukey_result.reject)):
        g1 = tukey_result._results_table.data[i+1][0]
        g2 = tukey_result._results_table.data[i+1][1]
        reject = tukey_result.reject[i]
        if reject:
            sig_matrix.loc[g1, g2] = 0
            sig_matrix.loc[g2, g1] = 0

    # Asignar letras
    current_letter = "a"
    for g in groups:
        if letters[g] == "":
            letters[g] = current_letter
            for h in groups:
                if sig_matrix.loc[g, h] == 1 and letters[h] == "":
                    letters[h] = current_letter
            current_letter = chr(ord(current_letter) + 1)

    return pd.DataFrame({"Grupo": list(letters.keys()), "Letra": list(letters.values())})
```

```python
#para Peso
letras_peso = tukey_to_letters(tukey_peso)
letras_peso
```

```python
# para Long
letras_long = tukey_to_letters(tukey_long)
letras_long
```

```python
# tabla completa
# col grupo
df_IV['Especie'] = df_IV['Especie'].astype(str)
df_IV['Tratamiento'] = df_IV['Tratamiento'].astype(str)
df_IV['Grupo'] = df_IV[['Especie','Tratamiento']].agg('_'.join, axis=1)
```

```python
#Tabla estadística base (medias, sd, n, se)
tabla_stats = (
    df_IV
    .groupby("Grupo")
    .agg(
        Peso_media=("Peso_g", "mean"),
        Peso_sd=("Peso_g", "std"),
        Peso_n=("Peso_g", "count"),
        Long_media=("Longitud_cm", "mean"),
        Long_sd=("Longitud_cm", "std"),
        Long_n=("Longitud_cm", "count")
    )
    .reset_index()
)

# Error estándar
tabla_stats["Peso_se"] = tabla_stats["Peso_sd"] / tabla_stats["Peso_n"]**0.5
tabla_stats["Long_se"] = tabla_stats["Long_sd"] / tabla_stats["Long_n"]**0.5
```

```python
# añadir letras sig Tukey
tabla_completa = (
    tabla_stats
    .merge(letras_peso.rename(columns={"Letra": "Letra_Peso"}), on="Grupo")
    .merge(letras_long.rename(columns={"Letra": "Letra_Longitud"}), on="Grupo")
)
```

```python
import scipy.stats as st

# IC 95% para Peso
tabla_completa["Peso_IC95_inf"] = tabla_completa["Peso_media"] - 1.96 * tabla_completa["Peso_se"]
tabla_completa["Peso_IC95_sup"] = tabla_completa["Peso_media"] + 1.96 * tabla_completa["Peso_se"]

# IC 95% para Longitud
tabla_completa["Long_IC95_inf"] = tabla_completa["Long_media"] - 1.96 * tabla_completa["Long_se"]
tabla_completa["Long_IC95_sup"] = tabla_completa["Long_media"] + 1.96 * tabla_completa["Long_se"]
```

```python
# Anova y p_values

modelo_peso = ols("Peso_g ~ Especie * Tratamiento", data=df_IV).fit()
anova_peso = sm.stats.anova_lm(modelo_peso, typ=2)
anova_peso
modelo_long = ols("Longitud_cm ~ Especie * Tratamiento", data=df_IV).fit()
anova_long = sm.stats.anova_lm(modelo_long, typ=2)
anova_long
pvals_peso = {
    "p_Especie_Peso": anova_peso.loc["Especie", "PR(>F)"],
    "p_Tratamiento_Peso": anova_peso.loc["Tratamiento", "PR(>F)"],
    "p_Interaccion_Peso": anova_peso.loc["Especie:Tratamiento", "PR(>F)"]
}

pvals_long = {
    "p_Especie_Long": anova_long.loc["Especie", "PR(>F)"],
    "p_Tratamiento_Long": anova_long.loc["Tratamiento", "PR(>F)"],
    "p_Interaccion_Long": anova_long.loc["Especie:Tratamiento", "PR(>F)"]
}
for k, v in pvals_peso.items():
    tabla_completa[k] = v

for k, v in pvals_long.items():
    tabla_completa[k] = v

```

```python
# tukey sobreboxplot
def agregar_letras_boxplot(ax, df, variable, letras_df):
    posiciones = {g: i for i, g in enumerate(df["Grupo"].unique())}

    for _, fila in letras_df.iterrows():
        grupo = fila["Grupo"]
        letra = fila["Letra"]
        x = posiciones[grupo]
        y = df[df["Grupo"] == grupo][variable].max()

        ax.text(x, y + 0.05*y, letra, ha='center', va='bottom',
                fontsize=14, color='black', fontweight='bold')
```

```python
#  figura combinada (Peso + Longitud 
def agregar_letras_boxplot(ax, df, variable, letras_df):
    posiciones = {g: i for i, g in enumerate(df["Grupo"].unique())}

    for _, fila in letras_df.iterrows():
        grupo = fila["Grupo"]
        letra = fila["Letra"]
        x = posiciones[grupo]
        y = df[df["Grupo"] == grupo][variable].max()

        ax.text(
            x,
            y + 0.05 * y,
            letra,
            ha='center',
            va='bottom',
            fontsize=14,
            color='black',
            fontweight='bold'
        )
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# --- Panel 1: Peso ---
ax1 = sns.boxplot(
    data=df_IV,
    x="Grupo",
    y="Peso_g",
    hue="Grupo",
    palette="magma",
    legend=False,
    ax=axes[0]
)

agregar_letras_boxplot(ax1, df_IV, "Peso_g", letras_peso)

axes[0].set_title("Peso por Grupo (INVITRO)", fontsize=16)
axes[0].tick_params(axis='x', rotation=45)

axes[0].text(
    0.5, 1.05,
    p_peso,
    ha='center',
    va='center',
    transform=axes[0].transAxes,
    fontsize=13
)

# --- Panel 2: Longitud ---
ax2 = sns.boxplot(
    data=df_IV,
    x="Grupo",
    y="Longitud_cm",
    hue="Grupo",
    palette="viridis",
    legend=False,
    ax=axes[1]
)

agregar_letras_boxplot(ax2, df_IV, "Longitud_cm", letras_long)

axes[1].set_title("Longitud por Grupo (INVITRO)", fontsize=16)
axes[1].tick_params(axis='x', rotation=45)

axes[1].text(
    0.5, 1.05,
    p_long,
    ha='center',
    va='center',
    transform=axes[1].transAxes,
    fontsize=13
)

plt.tight_layout()
plt.show()
plt.savefig("INVITRO_Figura_combinada_Peso_Longitud.jpg", dpi=300, bbox_inches='tight')
plt.close()
```

# EMMeans = Estimated Marginal Means  

```python
# ajustar el modelo
modelo_peso = ols("Peso_g ~ Especie * Tratamiento", data=df_IV).fit()
modelo_long = ols("Longitud_cm ~ Especie * Tratamiento", data=df_IV).fit()
```

```python
#combinaciones Especie × Tratamiento
import itertools


# Cuadrícula Especie × Tratamiento
grupos = list(itertools.product(df_IV["Especie"].unique(),
                                df_IV["Tratamiento"].unique()))
grid = pd.DataFrame(grupos, columns=["Especie", "Tratamiento"])

# EMMeans Peso
emmeans_peso = grid.copy()
pred_peso = modelo_peso.get_prediction(grid).summary_frame()
emmeans_peso["EMM_Peso"] = pred_peso["mean"]
emmeans_peso["EMM_Peso_IC95_inf"] = pred_peso["mean_ci_lower"]
emmeans_peso["EMM_Peso_IC95_sup"] = pred_peso["mean_ci_upper"]

# EMMeans Longitud
emmeans_long = grid.copy()
pred_long = modelo_long.get_prediction(grid).summary_frame()
emmeans_long["EMM_Long"] = pred_long["mean"]
emmeans_long["EMM_Long_IC95_inf"] = pred_long["mean_ci_lower"]
emmeans_long["EMM_Long_IC95_sup"] = pred_long["mean_ci_upper"]
```

```python
emmeans_peso["Grupo"] = emmeans_peso["Especie"] + "_" + emmeans_peso["Tratamiento"]
emmeans_long["Grupo"] = emmeans_long["Especie"] + "_" + emmeans_long["Tratamiento"]
```

```python
# Unir EMMeans con letras Tukey y tabla completa
tabla_completa = (
    tabla_completa
    .merge(
        emmeans_peso[[
            "Grupo",
            "EMM_Peso",
            "EMM_Peso_IC95_inf",
            "EMM_Peso_IC95_sup"
        ]],
        on="Grupo",
        how="left"
    )
    .merge(
        emmeans_long[[
            "Grupo",
            "EMM_Long",
            "EMM_Long_IC95_inf",
            "EMM_Long_IC95_sup"
        ]],
        on="Grupo",
        how="left"
    )
)
tabla_completa
```

```python
tabla_completa = tabla_completa.rename(columns={
    "EMM_Peso_x": "EMM_Peso",
    "EMM_Peso_IC95_inf_x": "EMM_Peso_IC95_inf",
    "EMM_Peso_IC95_sup_x": "EMM_Peso_IC95_sup",
    "EMM_Long_x": "EMM_Long",
    "EMM_Long_IC95_inf_x": "EMM_Long_IC95_inf",
    "EMM_Long_IC95_sup_x": "EMM_Long_IC95_sup"
})

tabla_completa = tabla_completa.drop(columns=[
    "EMM_Peso_y",
    "EMM_Peso_IC95_inf_y",
    "EMM_Peso_IC95_sup_y",
    "EMM_Long_y",
    "EMM_Long_IC95_inf_y",
    "EMM_Long_IC95_sup_y"
])
```

Intervalos de confianza (IC95%)
